```python
import async_jupyter as asj
import asyncio
loop = asj.get_event_loop()
```

```python
async def keep_printing():
    i = 0

    while True:
        print(i)
        i +=1
        await asyncio.sleep(2)

loop.create_task(keep_printing())
        
```
