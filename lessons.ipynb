```python
import numpy as np
from io import BytesIO
import PIL

import ipywidgets as widgets
def png_bytes(array):
    a = np.uint8(np.transpose(array, [1,0,2]))
    f = BytesIO()
    PIL.Image.fromarray(a).save(f, 'PNG')

    return f.getvalue()

def to_ipython_image(a):

    return widgets.Image(value=png_bytes(a),
    format='PNG', width=a.shape[1], height=a.shape[0])


ipython_image = to_ipython_image(np.ones([100, 100, 3], dtype=np.uint8) * 200)

display(ipython_image)
```

```python
import numpy as np
import io
from PIL import Image
import ipywidgets as widgets
from IPython.display import display

# Step 1: Create your NumPy array (e.g., random image)
array = np.random.randint(0, 255, (100, 100, 3), dtype=np.uint8)

# Step 2: Convert the NumPy array to a PNG bytes buffer
buffer = io.BytesIO()
Image.fromarray(array).save(buffer, format='PNG')
png_data = buffer.getvalue()

# Step 3: Create an ipywidget Image using the PNG data
img_widget = widgets.Image(
    value=png_data,  # raw PNG data
    format='png',    # specify the format
    width=300,       # optional: width in pixels
    height=300       # optional: height in pixels
)

# Step 4: Display the widget
display(img_widget)
```
